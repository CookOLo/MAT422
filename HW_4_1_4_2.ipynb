# 4.2 - Graph modeling

- **graph:** consists of both a set of objects, called nodes, and the connections between those nodes, called edges
  - Mathematically, a graph G is denoted as pair $G(V, E)$, where $V = {v_1}, {v_2}, ... {v_n}$ represents the set of nodes and $E = {e_1}, {e_2}, ... {e_m}$
  - Edges are also represented by their endpoints (nodes), so $e(v_1, v_2) or (v_1, v_2)

- When edges are undirected, nodes are connected both ways and are called undirected edges and this kind of graph is called an undirected graph (vice versa with directed edges with directed graphs)

- **path**: a sequence of edges where nodes and edges are distinct, $e_1(v_1, v_2)$, $e_2(v_2, v_3)$,..., $e_i(v_i, v_{i+1})$
- **cycle**: closed path
- **length of a path or cycle**: the number of edges traversed in the path or cycle

- For a directed graph, there are two types of degrees:
1) in-degrees
    - edges toward the node
2) out-degrees
    - edges away from the node

- **in-degree centrality**: describes the popularity of a node and its prominence or prestige
- **out-degree centrality**: describes the gregariousness of the node

**Theorem 4.2.3**: The Laplacian matrix L of has the following properties:
1. $L = D - M$, where $M$ is the adjacency matrix and $D$ is the diagonal degree matrix $D_{ii} = \sum_k E_{ik}$
2. $L = I_G I_G^T$
3. $L$ is symmtric positive semi-definite. All eigenvalues of $L$ are real and non-negative, and $L$ has a full set of $n$ real and orthogonal eigenvectors
4. Let $e = [1,..., 1]^T$. Then $Le = 0$. Thus 0 is the smallest eigenvalue and $e$ is the corresponding eigenvector
5. If the graph $G$ has c connected components, then $L$ has $c$ eigenvalues that is 0.
6. For any vector $x, x^T Lx = \sum_\{{i, j\}∈E} E_{ij}(x_i - x_j)^2$
7. For any vector $x$ and scalars
$(\alpha x + \beta e)^T L (\alpha x + \beta e) = \alpha^2 x^T L x.$

9. The problem

$$\min_{x \neq 0} \; x^T L x \quad \text{subject to} \quad x^T x = 1, \quad x^T e = 0$$

is solved when $x$ is the eigenvector corresponding to the second smallest eigenvalue $\lambda_2$ (the Fiedler vector) of the eigenvalue problem

$$Lx = \lambda x$$

```python
# Showing Laplacian Matrix
## L = D (degree matrix) − A (adjacency matrix)

import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

## Adjacency matrix for the path 0 - 1 - 2
A = np.array([
    [0, 1, 0],
    [1, 0, 1],
    [0, 1, 0]
])

####### Showcasing the path graph ############
G = nx.path_graph(3)  # vertices 0 - 1 - 2

pos = {0: (0, 0), 1: (1, 0), 2: (2, 0)}

nx.draw(G, pos, with_labels=True, node_color="lightblue",
        node_size=800, font_size=14, width=2)
plt.title("Path graph of 0 - 1 - 2")
plt.show()
#################################################


## Degree matrix: diagonal entries are the row sums of A
D = np.diag(A.sum(axis=1))

## Laplacian matrix formula
L = D - A
print("\nThe adjacency matrix of the path graph above is:\n", L)
```

```python
# Showing incidence matrix
## which vertices are connected to edges

import networkx as nx

####### Showcasing the path graph ############
G = nx.path_graph(3)  # vertices 0 - 1 - 2

pos = {0: (0, 0), 1: (1, 0), 2: (2, 0)}

nx.draw(G, pos, with_labels=True, node_color="lightblue",
        node_size=800, font_size=14, width=2)
plt.title("Path graph of 0 - 1 - 2")
plt.show()
#################################################

B_unoriented = nx.incidence_matrix(G).toarray()

print("\nIf the first column is e1(0, 1) and the second column is e2(1, 2), then this is the incidence matrix of the path graph above:\n", B_unoriented)
```
