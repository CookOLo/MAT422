# 3.2 - Continuity and differentiation
## 3.2.1 - Limits and continuity


**Definition 3.2.1** (Limits of a function). Let f : D → R be a real-valued
function on D ⊆ Rd. Then f is said to have a limit L ∈ R as x approaches
a if: for any ε > 0, there exists a δ > 0 such that |f (x) − L| < ε for all
$x ∈ D ∩ B_δ (a) \ \{a\}$. This is written as

$$\lim_{x \to a} f (x) = L.$$


```python
# Limit

import sympy as sp
x = sp.Symbol('x')

## lim[x -> 2] (3x^2 - 1)
y = 3 * (x * x) - 1
l = sp.limit(y, x, 2)
print("lim[x -> 2] (3x^2 - 1) ==", l)

## lim[x -> 2] ((x^2 - 4) /(x - 2))
y = (x * x - 4) / (x - 2)
l = sp.limit(y, x, 2)
print("lim[x -> 2] ((x^2 - 4) /(x - 2)) ==", l)
```

**Definition 3.2.2** (Continuous function). Let f : D → R be a real-valued
function on D ⊆ Rd. Then f is said to be continuous at a ∈ D if

$$\lim_{x \to a} f (x) = f (a).$$

**Definition 3.2.4** (Extremum). Suppose f : D → R is defined on a set
$D ⊆ R^d$. We say that f attains a maximum value $M$ at $z∗$ if $f (z∗) = M$
and $M ≥ f (x)$ for all $x ∈ D$. Similarly, we say $f$ attains a minimum value $m$ at $z_∗$ if $f (z_∗) = m$ and $m ≤ f (x)$ for all $x ∈ D$.

**Theorem 3.2.5** (Extreme value). Let $f : D → R$ be a real-valued, continuous
function on a nonempty, closed, bounded set $D ⊆ R^d$ . Then $f$ attains a maximum and a minimum on $D$.

```python
# Continuous Function Graph

import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**3.0 - 2.0 * x + np.cos(x)

fx1 = np.linspace(-2, 2, 1000)
plt.figure(figsize=(20, 5))
plt.plot(fx1, f(fx1))
plt.show()
```

## 3.2.2 - Derivatives
### 3.2.2.1 - Single-variable case

**Definition 3.2.6.** (Derivative). Let $f : D → R$ where $D ⊆ R$ and let $x_0 ∈ D$ be an interior point of $D$. The derivative of $f$ at $x_0$ is

$$f ′(x0) = \frac{df (x_0)}{dx} = \lim_{h→0} \frac{f (x_0 + h) − f (x_0)}{h},$$

provided the limit exists.

```python
# Derivatives

from sympy import *
x = Symbol('x')

## derivative of (x³ - 2x² + 5)
f = x**3 - 2 * x**2 + 5
print("The derivative of (x³ - 2x² + 5) is: ");
f.diff(x)
```

**Proposition 3.2.7.** Let f and g have derivatives at x and let α and β be
constants. The following results hold:

$$[αf (x) + βg(x)]′ = αf ′(x) + βg′(x).$$

**Theorem 3.2.9.** (Rolle). Let $f : [a, b] → R$ be a continuous function and assume that its derivative exists on $(a, b)$. If $f (a) = f (b)$, then there is $a < c < b$ such that $f ′(c) = 0$

**Theorem 3.2.10** (Mean value). Let $f : [a, b] → R$ be a continuous function
and assume that its derivative exists on $(a, b)$. Then there is $a < c < b$ such that

$$f(b) = f(a) + (b − a)f ′(c),$$

or put differently

$$\frac{f (b) − f (a)}{b − a} = f ′(c).$$

### 3.2.2.2 General Case

**Definition 3.2.11** (Partial derivative). Let $f : D → R^m$ where $D ⊆ R^d$ and let $x_0 ∈ D$ be an interior point of D. The partial derivative of $f$ at $x_0$ with respect to $x_i$ is

$$∂f (x0) ∂xi = lim h→0 f (x0 + hei) − f (x0) h,$$

provided the limit exists. If $\frac{∂f (x0)} {∂xi}$ exists and is continuous in an open ball around x0 for all $i$, then we say that $f$ continuously differentiable at $x_0$

```python
# Partial Derivative
from sympy import *

## partial derivative of sin(x)cos(y) + xy
x, y = symbols('x y', real=True)
f = sin(x) * cos(y) + x * y
print("The partial derivative of sin(x)cos(y) + xy is: ");
f.diff(x)
```

{∂xi}$ exists for all $i$, $j$. The
Jacobian of $f$ at $x_0$ is the $d × m$ matrix

For a real-valued function $f : D → R$, the Jacobian reduces to the row
vector

$$J_f (x_0) = ∇f (x_0)^T,$$

where the vector

$$∇f (x_0) = (\frac{∂f (x0)}{∂x1}, . . . , \frac{∂f (x0)}{∂xd})^T$$

is the gradient of $f$ at $x_0$.

**Theorem 3.2.15** (Chain rule). Let $f : D_1 → R^m$ , where $D_1 ⊆ R^d$ , and let $g : D_2 → R^p$ , where $D_2 ⊆ R^m$ . Assume that $f$ is continuously differentiable at $x_0$, an interior point of $D_1$, and that $g$ is continuously differentiable at $f(x_0)$, an interior point of $D_2$. Then

$$J_{g◦f(x0)} = J_g(f(x_0)) J_f(x_0)$$

as a product of matrices.

**Theorem 3.2.17** (Directional derivative from gradient). Let $f : D → R$
where $D ⊆ R^d$, let $x_0 ∈ D$ be an interior point of $D$, and let $v ∈ R^d$ be a unit vector.
Assume that f is continuously differentiable at $x_0$. Then the directional derivative of f at $x_0$ in the direction $v$ is given by

$$\frac{∂f (x0)} {∂v} = J_f (x_0) v = ∇f (x_0)^T v.$$

**Definition 3.2.18** (Second partial derivatives and Hessian). Let $f : D → R$
where $D ⊆ R^d$ and let $x_0 ∈ D$ be an interior point of $D$. Assume that $f$ is continuously differentiable in an open ball around $x_0$ . Then $∂f (x)/∂xi$ is itself a function of $x$ and its partial derivative with respect to xj, if it exists, is denoted by

$$\frac{∂^2f (x_0)}{∂x_j∂x_i} = \lim_{h→0} \frac{∂f (x_0 + he_j)/∂x_i − ∂f (x_0)/∂x_i} {h}.$$

**Theorem 3.2.19** (Symmetry of the Hessian). Let $f : D → R$ where $D ⊆ R^d$
and let $x_0 ∈ D$ be an interior point of $D$. Assume that $f$ is twice continuously differentiable at $x_0$. Then for all $i != j$,

$$\frac{∂^2f (x_0)} {∂x_j∂x_i} = \frac{∂^2f (x_0)} {∂x_i∂x_j}.$$

```python
# Chain Rule (same as any other derivative written in Python)
from sympy import *
x = Symbol('x')

## derivative of sin(x² + 3x) -- chain rule
f = sin(x**2 + 3 * x)
print("The derivative of sin(x² + 3x) is (using chain rule): ");
f.diff(x)
```

**Theorem 3.2.21** (Taylor). Let $f : D → R$ where $D ⊆ R$. Suppose $f$ has a $m$ times continuous derivative on $[a, b]$. Then

$$f(b) = f(a) + (b − a)f ′(a) + \frac{1}{2} (b − a)^2 f′′(a) + ... + \frac{(b − a)^{m−1}}{(m − 1)!} f^{(m−1)} (a) + R_m$$

where $R_m = \frac{(b−a)^m}{(m)!} f^{(m)} (a + θ(b − a))$ for some $0 < θ < 1$. In particular, for $m = 2$, we have

$$f (b) = f (a) + (b − a)f ′(a) + \frac{1}{2} (b − a)^2f ′′(ξ )$$

for some $a < ξ < b$.

**Theorem 3.2.22** (Multivariate mean value). Let $f : D → R$ where $D ⊆ R^d$.
Let $x_0 ∈ D$ and $δ > 0$ be such that $B_δ (x_0) ⊆ D$. If $f$ is continuously differentiable on $B_δ (x_0)$, then for any $x ∈ B_δ (x_0)$,

$$f (x) = f (x0) + ∇f (x0 + ξ p)T p,$$

for some $ξ ∈ (0, 1)$, where $p = x − x_0$.

**Theorem 3.2.23** (Multivariate Taylor). Let $f : D → R$ where $D ⊆ Rd$. Let
$x_0 ∈ D$ and $δ > 0$ be such that $B_δ (x_0) ⊆ D$. If f is three times continuously differentiable on $B_δ (x_0)$, then for any $x ∈ B_δ (x_0)$,

$$f (x) = f (x_0) + ∇f (x_0)^T p + \frac{1}{2} p^T H_f (x_0) p + O(‖p‖^3),$$

where $p = x − x0.$

If $f$ is twice continuously differentiable on $B_δ (x_0)$, then for any $x ∈ B_δ (x_0)$,

$$f (x) = f (x_0) + ∇f (x_0)^T p + \frac{1}{2} p^T H_f (x_0 + ξ p) p, $$

for some $ξ ∈ (0, 1)$, where $p = x − x_0$.
