## **1.2.1: Linear Spaces**

**Definition 1.2.1 (Linear subspace).** A linear subspace of V is a subset
U ⊆ V that is closed under vector addition and scalar multiplication.

```python
# Imports
import numpy as np
```

```python
def simpleSubspaceCheckOrigin(x, y):
  # Origin Check
  if(x == 0 and y == 0):
    return True;

  return False;


def simpleSubspaceCheckAdd(x, y, fixedX, fixedY):
  # Vector Addition Check
  if(fixedX != None):
    if(fixedX != x):
      return False;
  if(fixedY != None):
    if(fixedY != y):
      return False;

  return True;

def simpleSubspaceCheckMult(x, y, negX, negY):
  # Scalar Mult Check
  if(negX == True):
    if(y * x >= 0):
      return False;

  if(negY == True):
    if(x * y >= 0):
      return False;

  return True;

# Examples for if V is a subset of U = R^2
## If U was smaller, defined space, then these functions won't work, but it's trivial to make a check for going out of bounds
## These specific functions were created for the simple demonstration of `U = R^2`

# Representing Vectors {x, -y} in V are a subspace of U = R^2

broke = False;
addCheck = False;
zeroCheck = False;
multCheck = False;
print("Testing whether V with vectors {x, -(y^2)} is a subspace of U = R^2\n");
for x in range(-999, 999): # Max nums are limited to -999/999 for time sakes (runs slower on bigger values)
  for y in range(-999, 999):
    if(zeroCheck == False and simpleSubspaceCheckOrigin(x, -(y * y)) == True):
        zeroCheck = True;
        broke = True;
        break;
  if broke == True:
      break;

if zeroCheck == False:
  print("V is not a subspce of U because there is no Vector 0 in V\n\n");

for x in range(-999, 999): # Max nums are limited to -999/999 for time sakes (runs slower on bigger values)
  for y in range(-999, 999):
    if(addCheck == False and simpleSubspaceCheckAdd(x, -(y * y), None, None) == True):
        addCheck = True;
        broke = True;
        break;
  if broke == True:
    break;

if addCheck == False:
  print("V is not a subspce of U because it is not enclosed in vector addition\n\n");

for x in range(-999, 999): # Max nums are limited to -999/999 for time sakes (runs slower on bigger values)
  for y in range(-999, 999):
    if(multCheck == False and simpleSubspaceCheckMult(x, -(y * y), False, True) == True):
        multCheck = True;
        broke = True;
        break;
  if broke == True:
      break;

if multCheck == False:
  print("V is not a subspce of U because it is not enclosed under scalar multiplication\n\n");

if multCheck == True and addCheck == True and zeroCheck == True:
  print("V is a subspace of U!\n");

# Representing Vectors {1, y} in V are a subspace of U = R^2

broke = False;
addCheck = False;
zeroCheck = False;
multCheck = False;
print("Testing whether V with vectors {1, y} is a subspace of U = R^2\n");
for x in range(-999, 999): # Max nums are limited to -999/999 for time sakes (runs slower on bigger values)
  for y in range(-999, 999):
    if(zeroCheck == False and simpleSubspaceCheckOrigin(1, y) == True):
        zeroCheck = True;
        broke = True;
        break;
  if broke == True:
      break;

if zeroCheck == False:
  print("V is not a subspce of U because there is no Vector 0 in V\n\n");

for x in range(-999, 999): # Max nums are limited to -999/999 for time sakes (runs slower on bigger values)
  for y in range(-999, 999):
    if(addCheck == False and simpleSubspaceCheckAdd(x, y, 1, None) == True):
        addCheck = True;
        broke = True;
        break;
  if broke == True:
    break;

if addCheck == False:
  print("V is not a subspce of U because it is not enclosed in vector addition\n\n");

for x in range(-999, 999): # Max nums are limited to -999/999 for time sakes (runs slower on bigger values)
  for y in range(-999, 999):
    if(multCheck == False and simpleSubspaceCheckMult(x, y, False, False) == True):
        multCheck = True;
        broke = True;
        break;
  if broke == True:
      break;

if multCheck == False:
  print("V is not a subspce of U because it is not enclosed under scalar multiplication\n\n");

if multCheck == True and addCheck == True and zeroCheck == True:
  print("V is a subspace of U!\n");

# Representing Vectors {x, 0} in V are a subspace of U = R^2

broke = False;
addCheck = False;
zeroCheck = False;
multCheck = False;
print("Testing whether V with vectors {x, 0} is a subspace of U = R^2\n");
for x in range(-999, 999): # Max nums are limited to -999/999 for time sakes (runs slower on bigger values)
  for y in range(-999, 999):
    if(zeroCheck == False and simpleSubspaceCheckOrigin(x, y) == True):
        zeroCheck = True;
        broke = True;
        break;
  if broke == True:
      break;

if zeroCheck == False:
  print("V is not a subspce of U because there is no Vector 0 in V\n\n");

for x in range(-999, 999): # Max nums are limited to -999/999 for time sakes (runs slower on bigger values)
  for y in range(-999, 999):
    if(addCheck == False and simpleSubspaceCheckAdd(x, y, None, None) == True):
        addCheck = True;
        broke = True;
        break;
  if broke == True:
    break;

if addCheck == False:
  print("V is not a subspce of U because it is not enclosed in vector addition\n\n");

for x in range(-999, 999): # Max nums are limited to -999/999 for time sakes (runs slower on bigger values)
  for y in range(-999, 999):
    if(multCheck == False and simpleSubspaceCheckMult(x, y, False, False) == True):
        multCheck = True;
        broke = True;
        break;
  if broke == True:
      break;

if multCheck == False:
  print("V is not a subspce of U because it is not enclosed under scalar multiplication\n\n");

if multCheck == True and addCheck == True and zeroCheck == True:
  print("V is a subspace of U!\n");

```

**Definition 1.2.5** (Linear independence). A list of vectors `u1, . . . ,um` is
linearly independent if none of them can be written as a linear combination of the others

```python
vectors = [[2, 3], [6, 9], [4, 9]];

def linDep(vector1, vector2):
  if(vector1[0] > vector2[0]):
    if((vector1[0] / vector2[0]) == (vector1[1] / vector2[1])):
      return True;
    else:
      return False;
  else:
    if ((vector2[0] / vector1[0]) == (vector2[1] / vector1[1])):
      return True;
    else:
      return False;
  return False;

# Testing
print("Testing whether Vector [2, 3] is linearally independent or dependent from Vector [6, 9]...\n")
if(linDep(vectors[0], vectors[1]) == True):
  print("It's linearaly dependent!\n\n");
else:
  print("It's linearaly independent!\n\n");

print("Testing whether Vector [2, 3] is linearally independent or dependent from Vector [4, 9]...\n")
if(linDep(vectors[0], vectors[2]) == True):
  print("It's linearaly dependent!\n\n");
else:
  print("It's linearaly independent!\n\n");
```


**Definition 1.2.7** (Basis of a space). Let U be a linear subspace of V .
A basis of U is a list of vectors `u1, . . . ,um` in U that:
(1) span U, that is,
U = span(`u1, . . . ,um`);
(2) are linearly independent.

**Theorem 1.2.8** (Dimension theorem). Let U be a linear subspace of V. Any basis of U always has the same number of elements. All bases of U have the same length, that is, the same number of elements. We call this number the dimension of
U and denote it `dim(U)`.

Definition 1.2.11. A list of vectors `{u1, . . . , um}` is orthonormal if the ui’s are pairwise orthogonal and each has norm 1, that is, for all i and all j != i,

`〈ui, uj〉 = 0` and `‖ui‖ = 1`

```python
def orthonormalCheck(vector1, vector2):
  # Checking if norm equals 1
  norm1 = 0;
  norm2 = 0;

  for i in range(len(vector1)):
    vector1[i] = vector1[i] * vector1[i];
    vector2[i] = vector2[i] * vector2[i];

  norm1 = np.sqrt(np.sum(vector1));
  norm2 = np.sqrt(np.sum(vector2));

  if norm1 != 1 or norm2 != 1:
    return False;

  # Checking if inner product equals 0
  inProd = np.dot(vector1, vector2)

  if(inProd != 0):
    return False;

  return True;



# Vectors
vec1 = [-2, 6, 9]
vec2 = [1, 0, 0]
vec3 = [0, 1, 0]


if(orthonormalCheck(vec1, vec2) == False):
    print("Vector [-2, 6, 9] and Vector [1, 0, 0] are NOT orthonormal!");
else:
    print("Vector [-2, 6, 9] and Vector [1, 0, 0] are orthonormal!");

if(orthonormalCheck(vec2, vec3) == False):
    print("Vector [1, 0, 0] and Vector [0, 1, 0] are NOT orthonormal!");
else:
    print("Vector [1, 0, 0] and Vector [0, 1, 0] are orthonormal!");
```

## **1.2.3 - Gram-Schmidt process**

Theorem 1.2.20 (Gram–Schmidt). Let `a1, . . . ,am` in `Rn` be linearly independent. Then there exist an orthonormal basis `q1, . . . ,qm` of `span(a1, . . . , am)`


image.png

```python
def gramSchmidt(vector):
  repeat = 0;
  for i in range(len(vector)):
    if i == 0:
      repeat += 1;
      continue;

    back = repeat;
    for j in range(repeat):
      copyVec = vector[i].copy();
      proj = (np.dot(copyVec, vector[i - back]) / np.dot(vector[i - back], vector[i - back]));
      for l in range(len(vector[i - back])):
        copyVec[l] = proj * vector[i - back][l];
      back -= 1;


    for k in range(len(vector[i])):
      vector[i][k] = float(vector[i][k] - copyVec[k])

  return vector;

gramVec = [[1, -1, 1], [1, 0, 1], [1, 1, 2]]
print("The gramSchmidt of vectors [[1, -1, 1], [1, 0, 1], [1, 1, 2]] is: " + str(gramSchmidt(gramVec)));

gramVec = [[1, 2, -2], [4, 3, 2], [1, 2, 1]]
print("The gramSchmidt of vectors [[1, 2, -2], [4, 3, 2], [1, 2, 1]] is: " + str(gramSchmidt(gramVec)));
```

**Definition 1.2.21** (Eigenvalues and eigenvectors). Let A ∈ Rd×d be a
square matrix. Then λ ∈ R is an eigenvalue of A if there exists a non-zero
vector x != 0 such that

Ax = λx.

```python
matrix = [[0, 3], [2, 1]]

eval, evect = np.linalg.eig(matrix)
print("The eigenvalues of the matrix [[0, 3], [2, 1]] are: " + str(eval))
print("The eigenvectors of the matrix [[0, 3], [2, 1]] are: " + str(evect) + "\n")

# Shows no real eigenvalues
matrix = [[0, 1], [-1, 0]]

eval, evect = np.linalg.eig(matrix)
print("The eigenvalues of the matrix [[1, 0], [-1, 0]] are: " + str(eval))
print("The eigenvectors of the matrix [[1, 0], [-1, 0]] are: " + str(evect))
```

**Theorem 1.2.26** (The spectral theorem for symmetric matrices). An n × n symmetric matrix A has the following properties:

• A has n real eigenvalues, counting multiplicities.

• If λ is an eigenvalue of A with multiplicity k, then the eigenspace for λ is k-dimensional.

• The eigenspaces are mutually orthogonal, in the sense that eigenvectors corresponding to different eigenvalues are orthogonal.


• A is orthogonally diagonalizable.
