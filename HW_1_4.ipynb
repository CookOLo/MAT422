## 1.4.1 Singular Value Decomposition

$$A = U ∑ V^T$$

- $A$ is a non-symmtric matrix
- $U$ are the left singular vectors ($AA^T$)/orthogonal matrix
- $∑$ is the rectangular diagonal
- $V$ are the right singular vectors ($A^T A$)/orthogonal matrix

```python
nonsymMatrix = [[3, 2, 2], [2, 3, -2]];

# python has a method to calculate svd
u, s, vt = np.linalg.svd(nonsymMatrix, full_matrices=True);


print("This the U, sigma, and V^T of non-symmetrical matrix A which is [[3, 2, 2], [2, 3, -2]]");
print("U:\n", u);

print("\nSigma:\n", s);

print("\nV^T:\n", vt);
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns; sns.set()
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris # flower samples iwth 4 features

iris = load_iris()

pca = PCA(2) # Project from 4 to 2 dimensions
projected = pca.fit_transform(iris.data)

# Graph it out
plt.scatter(projected[:, 0], projected[:, 1], c=iris.target, edgecolor='none', alpha=0.5, cmap=plt.cm.get_cmap('Spectral', 10))
plt.xlabel('c1')
plt.ylabel('c2')

plt.colorbar();
```
