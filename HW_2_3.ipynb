# 2.3 - Independent variables and random samples
## 2.3.1 - Joint probabilty distributions
### 2.3.1. - Two discrete random variables
**Definition 2.3.1.** Let X and Y be two discrete random variable’s defined
on the sample space S of an experiment. The joint probability mass
function $p(x, y)$ is defined for each pair of numbers $(x, y)$ by

$$p(x, y) = P(X = x\;and\;Y = y).$$

It must be the case that $p(x, y) ≥ 0$ and $∑_x∑_y p(x, y) = 1.$

**Definition 2.3.2.** The marginal probability mass function of X, de-
noted by $p_X (x)$, is given by

$$p_X (x) = ∑_{y:p(x,y)>0} p(x, y)$$ for each possible value x.


Similarly, the marginal probability mass function of Y is

$$p_Y (y) = ∑_{x:p(x,y)>0} p(x, y)$$ for each possible value y.
### 2.3.1.2 - Two continuous random variables

**Definition 2.3.3.** Let $X$ and $Y$ be continuous random variables. A joint
probability density function $f(x, y)$ for these two variables is a function
satisfying $f(x, y) ≥ 0$ and $∫_{-∞}^∞∫_{-∞}^∞ f (x, y)dxdy = 1$. Then for any two-dimensional set A,

$$P[(X, Y) ∈ A] = ∫∫_A f(x, y)dxdy.$$
In particular, if A is the two-dimensional rectangle ${(x, y) : a ≤ x ≤ b, c ≤
y ≤ d}$ then

$$P[(X, Y ) ∈ A] = P(a ≤ X ≤ b, c ≤ Y ≤ d) = ∫_a^b ∫_c^d f (x, y)dydx.$$

**Definition 2.3.4.** The marginal probability density functions of $X$
and $Y$ , denoted by $f_X(x)$ and $f_Y(y)$, respectively, are given by
$$f_X (x) = ∫^∞_{−∞} f (x, y)dy for −∞ < x < ∞$$
$$f_Y (y) = ∫^∞_{−∞} f (x, y)dx for −∞ < y < ∞$$

### 2.3.1.3 - Independent random variables

**Definition 2.3.5.** Two random variables X and Y are said to be independent if for every pair of x and y values,
$$p(x, y) = p_X (x) · p_Y (y)$$ when X and Y are discrete

or

$$f (x, y) = f_X (x) · f_Y (y)$$ when X and Y are continuous.

**Definition 2.3.7.** If $X_1, X_2, . . . ,X_n$ are all discrete random variables, the joint pmf of the variables is the function

$$p (x1, x2, . . . , xn) = P (X1 = x1, X2 = x2, . . . , X n = xn).$$

If the variables are continuous, the joint pdf of $X_1, . . ., X_n$ is the function $f (x1, x2, . . . , xn)$ such that for any $n$ intervals $[a1, b1] , . . . , [an, bn]$,

$$P (a_1 ≤ X_1 ≤ b_1, . . . , a_n ≤ X_n ≤ b_n) = ∫^{b_1}_{a_1} . . . ∫^{b_n}_{a_n} f(x_1, . . ., x_n) dx_n . . . dx_1$$

**Definition 2.3.8.** The random variables $X_1, X_2, . . ., X_n$ are said to be independent if for every subset $X{i_1}, X{i_2}, . . ., X_{i_k}$ of the variables (each pair, each triple, and so on), the joint pmf or pdf of the subset is equal to the product of the marginal pmf’s or pdf’s.

```python
# With independent, random variables, joint probability distribution can be done (as a scatter plot)
np.random.seed(76)
n = 215325 # sample size
x = np.random.randn(n)
y = -np.random.randn(n)**2 # left-skewed, negative

## plotting
sns.set
sns.jointplot(x=x, y=y, kind = 'scatter', marginal_ticks=True)
plt.show()
```

## 2.3.2 - Correlation and dependence
### 2.3.2.1 - Correlation for random variables

**Definition 2.3.9.** Let X and Y be jointly distributed random variables
with pmf $p(x, y)$ or pdf $f(x, y)$ according to whether the variables are discrete or continuous. The covariance between two random variables X and Y

**Definition 2.3.10.** The correlation coefficient of $X$ and $Y$, denoted by
$Corr(X, Y )$, $ρ_X,Y$, or just $ρ$, is defined by

$$ρX,Y = \frac{Cov(X, Y)}{σ_X · σ_Y}$$

**Proposition 2.3.11.** Correlation coefficients have the following properties:
1. If $X$ and $Y$ are independent, then $ρ = 0$, but $ρ = 0$ does not imply independence.
2. $|ρ| ≤ 1$, $ρ = 1$ or $−1$ if $Y = aX + b$ for some numbers $a$ and $b$ with $a = 0$.

### 2.3.2.2 - Correlation for samples

**Proposition 2.3.12.** The correlation coefficient for samples has the following properties:
- $r_{xy} = \frac{s_{xy}}{s_x s_y}$ where the sample covariance

$$s_{xy} = \frac{1}{n − 1} \sum^n_{i=1} (x_i − \bar{x})(y_i − \bar{y})$$

and the sample standard deviation

$$s_x = \sqrt{\frac{1}{n − 1} \sum^n_{i=1} (x_i − \bar{x})^2}$$
- If $y = ax + b$, then $r={xy} = 1$ or $−1$ if a is positive or negative.

```python
# Using the previous random x and y values, the coefficient correlation is:

pearson, _ = pearsonr(x, y)
print(pearson)
print("\nThat basically means those values have low correlation")

# High Positive Correlation
x = [2, 3, 4, 5, 6, 7]
y = [3, 4, 5, 6, 7, 8]
pearson, _ = pearsonr(x, y)
print("\nBut if has high positive correlation with x values of [2, 3, 4, 5, 6, 7] and y values of [3, 4, 5, 6, 7, 8], then it's correlation would be: " + str(pearson))

## plotting that high correlation
sns.set
sns.jointplot(x=x, y=y, kind = 'scatter', marginal_ticks=True)
plt.show()

# High Negative Correlation
x = [-2, -3, -4, -5, -6, -7]
y = [3, 4, 5, 6, 7, 8]
pearson, _ = pearsonr(x, y)
print("\nBut if has high negative correlation with x values of [-2, -3, -4, -5, -6, -7] and y values of [-3, -4, -5, -6, -7, -8], then it's correlation would be: " + str(pearson))

## plotting that high correlation
sns.set
sns.jointplot(x=x, y=y, kind = 'scatter', marginal_ticks=True)
plt.show()
```

## 2.3.3 - Random Samples
### 2.3.3.1 - Random samples

**Definition 2.3.13.** The random variables $X_1, X_2, . . ., X_n$ are said to form a (simple) random sample of size $n$ if:
1. The $X_i’s$ are independent random variables.
2. Every $X_i$ has the same probability distribution.

**Proposition 2.3.14.** Let $X$1, X$2, . . ., X_n$ be a random sample from a distribution with mean value $μ$ and standard deviation $σ$. Then:
1. $E(\bar{X}) = μ_{\bar{X}} = μ;$
2. $V (\bar{X}) = σ^2_{\bar{X}} = \frac{σ^2}{n}$ and $σ_X = \frac{σ}{\sqrt{n}}.$
In addition, with $T_o = X_1 + · · · + X_n$ (the sample total), $E(T_o) = nμ, V (T_o) = nσ^2, and σ_{T_o} = \sqrt{n}σ$.

### 2.3.3.2 - The central limit theorem

**Theorem 2.3.15** (The central limit theorem (CLT)). Let $X_1, X_2, . . . , X_n$ be a random sample from a distribution with mean $μ$ and variance $σ^2$. Then if $n$ is sufficiently large, $\bar{X}$ has approximately a normal distribution with $μ\bar{X} = μ$ and $σ^2_{\bar{x}} = \frac{σ}{2/n}$, and $T_o$ also has approximately a normal distribution with $μ_{T_o} = nμ, σ^2_{T_o} = nσ^2$. The larger the value of $n$, the better the approximation.

```python
## showing normal distribution with not a lot of values

sample_size=10
rate = 0.5

df10 = pd.DataFrame()

for i in range(1, 20):
    exponential_sample = np.random.exponential(1/rate, sample_size)
    col = f'sample {i}'
    df10[col] = exponential_sample

df10_sample_means = pd.DataFrame(df10.mean(), columns=['Sample means'])
sns.displot(df10_sample_means, kde=True);

## showing normal distribution with a lot of values

sample_size=1000
rate = 2.5

df1000 = pd.DataFrame()

for i in range(1, 80):
    exponential_sample = np.random.exponential(1/rate, sample_size)
    col = f'sample {i}'
    df1000[col] = exponential_sample

df1000_sample_means = pd.DataFrame(df1000.mean(), columns=['Sample means'])
sns.displot(df1000_sample_means, kde=True);
```
